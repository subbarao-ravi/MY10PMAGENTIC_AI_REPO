```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
GOOGLE_API_KEY = os.getenv('GOOGLE_API_KEY')
```

```python
myllm=ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
```

```python
myllm.invoke("hi how are you?").content[0]["text"]
```

```python
while True:
      user_input = input("you: ")
      if user_input.lower() in ["exit", "quit"]:
            print("exiting the chat. good bye")
            break
      response = myllm.invoke(user_input).content[0]["text"]
      print(f"AI: {response}")
```
